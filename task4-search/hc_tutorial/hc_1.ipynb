Hill Climbing Algorithm
------------------------
- This program simulates a simple hill-climbing algorithm to match my hard-coded text/string.

```python
import random
import string

# word = "Hello World"
word = "Tony will conquer Foundations of AI unit"

total_itrs = 0
```

```python
def generate_random_solution(length = len(word)):
    return [random.choice(string.printable) for _ in range(length)]
```

```python
def evaluate(solution):
    target = list(word)
    diff = 0
    for i in range(len(target)):
        s = solution[i]
        t = target[i]
        diff += abs(ord(s) - ord(t))
    return diff
```

```python
def mutate_solution(solution):
    global total_itrs
    index = random.randint(0, len(solution) - 1)
    solution[index] = random.choice(string.printable)
    total_itrs += 1
    return total_itrs
```

```python
best = generate_random_solution()
best_score = evaluate(best)
```

```python
while True:
    print('Best score so far: ', best_score, '\tSolution: ', "".join(best), '\ton iteration: ', total_itrs)

    if best_score == 0:
        print('Total Iterations: ', total_itrs)
        total_itrs = 0
        break

    new_solution = list(best)
    itr = mutate_solution(new_solution)
    score = evaluate(new_solution)
    if score < best_score:
        best = new_solution
        best_score = score
```

Observation:
- The algorithm shows the 'Plateau' problem during the last iterations.
